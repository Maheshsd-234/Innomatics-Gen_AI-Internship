```python
#1. User Login Check
username="admin"
password="1234"
user=input("Enter the username:")
pa=input("Enter the password:")
if(user==username and pa==password):
    print("Login successful")
else:
    print("Invalid credentials")
```

```python
#2. Pass / Fail Analyzer
marks = [45, 78, 90, 33, 60]
pa=0
fa=0
for i in marks:
    if i>=50:
        pa=pa+1
    else:
        fa=fa+1
print("The number of students passed: ",pa)
print("The number of students failed: ",fa)
```

```python
#3.Simple Data Cleaner
names = [" Alice ", "bob", " CHARLIE "]
j=0
for i in names:
    names[j] = (i.lower()).strip()
    j =  j+1
print(names)
```

```python
#4.Message Length Analyzer
messages = ["Hi", "Welcome to the platform", "OK"]
for i in messages:
    n = len(i)
    print("The  message ",i," has length ", n)
    if n>10:
        print("It contains more than 10 characters")    
```

```python
#5.Error Message Detector
logs = ["INFO", "ERROR", "WARNING", "ERROR"]
er=0
for i in logs:
    if i=="ERROR":
        er=er+1
print('The error count in logs is :',er)
```
